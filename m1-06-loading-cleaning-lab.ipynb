## TASK 1

```python
import pandas as pd

tickets=pd.read_csv("data_safe_copy.csv")

if len(tickets) >= 500:
    print("There are at least 500 rows")
else: 
    print("Less than 500 rows")

tickets.head()
```

```python
tickets.info()
```

## TASK 2

```python
tickets["opened_at"]=pd.to_datetime(tickets["opened_at"])
tickets.info()
```

```python
tickets['category'] = tickets['category'].astype(str).str.strip().str.lower()
tickets['priority'] = tickets['priority'].astype(str).str.strip().str.lower()

print("Unique Categories:", tickets['category'].unique())
print("Unique Priorities:", tickets['priority'].unique())
```

## TASK 3

```python
tickets['resolution_minutes'] = pd.to_numeric(tickets['resolution_minutes'], errors='coerce')
missing_count = tickets['resolution_minutes'].isna().sum()

print("Missing count':",  missing_count)

print(tickets['resolution_minutes'].info())
```

## TASK 4

```python
tickets_clean = tickets.dropna(subset=["resolution_minutes"]).copy()

rows = len(tickets)
rows_clean = len(tickets_clean)
negative_count = (tickets_clean["resolution_minutes"] < 0).sum()

print("Rows before cleaning:", rows)
print("Rows after cleaning: ", rows_clean)
print("Negative resolution minutes found: ", negative_count)
print("Final categories: ", tickets_clean['category'].unique())
print("Final priorities: ", tickets_clean['priority'].unique())
```

## TASK 5

```python
summary = {
    "total_records": len(tickets),
    "clean_records": len(tickets_clean),
    "missing_resolution": tickets['resolution_minutes'].isna().sum(),
    "avg_resolution": tickets_clean['resolution_minutes'].mean(),
    "max_resolution": tickets_clean['resolution_minutes'].max()
}

print(summary)

is_valid = (summary["clean_records"] + summary["missing_resolution"]) == summary["total_records"]
print("Verify: Clean + Missing == Total -->", is_valid)
```
